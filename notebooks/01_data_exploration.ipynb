```python
df=df[["Score","Text"]]
df.head()
```

```python
def map_sentiment(Score):
    if Score <=2:
        return "Negative"
    elif Score ==3:
        return "Neutral"
    else:
        return "Positive"
df["sentiment"]=df["Score"].apply(map_sentiment)
```

```python
df.head()
```

```python
df["Text"].iloc[6]
```

```python
df["review_length"]=df["Text"].apply(len)
df["review_length"].describe()
```

```python
df["sentiment"].value_counts(normalize=True)
```

```python
import re

def clean(text):
    text=text.lower()

    text=re.sub(r"<.*?>","",text)

    text= re.sub(r"http\S+|www\S+", "", text)

    text = re.sub(r"[^a-z\s]", "", text)

    text = re.sub(r"\s+", " ", text).strip()
    
    return text
```

```python
df["clean_review"]=df["Text"].apply(clean)
df[["Text","clean_review"]].tail()
```

```python
import nltk
nltk.download("stopwords")
nltk.download("wordnet")
```

```python
from nltk.corpus import stopwords
stop_words= set(stopwords.words("english"))

def tok_and_remove_stopwords(text):
    tokens=text.split()
    tokens= [word for word in tokens if word not in stop_words]
    return tokens
```

```python
df["tokens"]=df["clean_review"].apply(tok_and_remove_stopwords)
```

```python
df.head()
```

```python
from nltk.stem import  WordNetLemmatizer

lemmatizer= WordNetLemmatizer()

def lemm_tokens(tokens):
    return[lemmatizer.lemmatize(token) for token in tokens]
```

```python
df["lemmatized_tokens"]=df["tokens"].apply(lemm_tokens)
```

```python
df[["tokens","lemmatized_tokens"]].tail()
```

```python
df["proccessed_text"]=df["lemmatized_tokens"].apply(lambda x:" " .join(x) )
df[["proccessed_text","Text"]].head()
```

```python
from sklearn.model_selection import train_test_split

x=df["proccessed_text"]
y=df["sentiment"]

X_train,X_test,y_train,y_test = train_test_split(
    x,
    y,
    test_size=0.25,
    stratify=y,
    random_state=42
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf= TfidfVectorizer(
    max_features=20000,
    ngram_range=(1,2),
    min_df=5,
    max_df=0.9,              
    sublinear_tf=True 
)
X_train_Tfid=tfidf.fit_transform(X_train)
X_test_Tfid=tfidf.transform(X_test)
```

```python
type(sklearn.feature_extraction.text)
```

```python
from sklearn.linear_model import LogisticRegression
log=LogisticRegression(
    max_iter=100,
    class_weight ="balanced"
)
log.fit(X_train_Tfid,y_train)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

y_pred= log.predict(X_test_Tfid)
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
Cvect= CountVectorizer(
    max_features=20000,
    ngram_range=(1,2),
    min_df=5
)
X_train_Count=Cvect.fit_transform(X_train)
X_test_Count=Cvect.fit_transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(
    max_iter=100,
    class_weight="balanced"
    

)
model.fit(X_train_Count,y_train)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
y_pred_count=model.predict(X_test_Count)
print(confusion_matrix(y_test,y_pred_count),classification_report(y_test,y_pred_count))
```

```python
from sklearn.svm import LinearSVC
svm_model=LinearSVC(
    class_weight="balanced"
)
svm_model.fit(X_train_Tfid,y_train)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
y_pred_svm=svm_model.predict(X_test_Tfid)
print(classification_report(y_test,y_pred_svm))

print(classification_report(y_test,y_pred_svm))
```

```python
from sklearn.svm import LinearSVC

final_svm = LinearSVC(
    class_weight="balanced",
    random_state=42
)
final_svm.fit(X_train_Tfid,y_train)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
y_pred_final=final_svm.predict(X_test_Tfid)
print (confusion_matrix(y_test,y_pred_final))
print (classification_report(y_test,y_pred_final))
```
